```python
from typing import List
```

Next Greater Element I

```python
class Solution:
    def nextGreaterElement(self, nums1: List[int], nums2: List[int]) -> List[int]:
        n = len(nums2)
        st = []
        d = {val:idx for idx,val in enumerate(nums2)}
        next_greater = [-1 for _ in range(n)]
        for i in range(n-1,-1,-1):
            while st and nums2[st[-1]]<=nums2[i]:
                st.pop()
            if st:
                next_greater[i] = st[-1]
            
            st.append(i)

        return [nums2[next_greater[d[i]]] if next_greater[d[i]]!=-1 else -1 for i in nums1]
            
```

Next Greater Element II

```python
class Solution:
    def nextGreaterElements(self, nums: List[int]) -> List[int]:
        n = len(nums)
        st = []
        next_greater = [-1 for _ in range(n)]
        for i in range(2*n-1,-1,-1):
            while st and st[-1]<=nums[i%n]:
                st.pop()
            if st:
                next_greater[i%n]=st[-1]
            st.append(nums[i%n])
        return next_greater
```

Next Smaller Element

```python
class Solution:
    def nextSmallerElements(self, nums: List[int]) -> List[int]:
        n = len(nums)
        st = []
        next_smaller = [-1 for _ in range(n)]
        for i in range(n-1,-1,-1):
            while st and st[-1]>=nums[i]:
                st.pop()
            if st:
                next_smaller[i]=st[-1]
            st.append(nums[i])
        return next_smaller
    
Solution().nextSmallerElements( [4, 8, 5, 2, 25])
```

Number of Greater Elements to the Right

```python
class Solution:
    def count_NGE(self, arr, indices):
        n = len(arr)
        st = []
        next_greater = [0 for _ in range(n)]
        for i in range(n-1,-1,-1):
            while st and arr[st[-1]]<=arr[i]:
                st.pop()
            if st:
                next_greater[i]+=(1+next_greater[st[-1]])
            st.append(i)
        return next_greater
    
Solution().count_NGE(arr = [3, 4, 2, 7, 5, 8, 10, 6], indices = [0, 5])
```
